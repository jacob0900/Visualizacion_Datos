```python
import geopandas as gpd
import plotly.express as px

# ── 1. Leer shapefile ─────────────────────────────────────────────────────────
gdf = gpd.read_file(
    r"C:\Users\jacob\OneDrive\Escritorio\MAPAS\Servicio-199\GDB_Lineas_Limitrofes_ET_Diciembre_2025_shp\Depto.shp"
)

# ── 2. Reproyectar a WGS84 (EPSG:4326) ───────────────────────────────────────
gdf = gdf.to_crs("EPSG:4326")

# ── 3. Convertir a GeoJSON ────────────────────────────────────────────────────
geojson = gdf.__geo_interface__

# ── 4. Mapa coroplético con plotly.express ────────────────────────────────────
fig = px.choropleth_mapbox(
    gdf,
    geojson=geojson,
    locations=gdf.index,
    color="DeArea",
    hover_name="DeNombre",
    hover_data={"DeArea": ":.2f", "DeNorma": True, "DeCodigo": True},
    color_continuous_scale="Viridis",
    mapbox_style="open-street-map",
    zoom=4,
    center={
        "lat": gdf.geometry.centroid.y.mean(),
        "lon": gdf.geometry.centroid.x.mean(),
    },
    opacity=0.65,
    labels={"DeArea": "Área (km²)", "DeNorma": "Norma", "DeCodigo": "Código"},
    title="Departamentos de Colombia — Diciembre 2025",
)

fig.update_layout(
    margin={"r": 0, "t": 50, "l": 0, "b": 0},
    coloraxis_colorbar=dict(
        title="Área (km²)",
        thickness=15,
        len=0.6,
    ),
)

fig.show()
```

```python
import geopandas as gpd
import plotly.express as px

def mapa_departamento(nombre_departamento: str):
    """
    Genera un mapa coroplético interactivo de los municipios de un departamento colombiano.
    
    Parámetros:
    -----------
    nombre_departamento : str
        Nombre del departamento. Ej: "Atlántico", "Cundinamarca", "Antioquia"
    """
    
    # ── 1. Leer shapefile ─────────────────────────────────────────────────────
    RUTA_SHP = r"C:\Users\jacob\OneDrive\Escritorio\MAPAS\Servicio-199\GDB_Lineas_Limitrofes_ET_Diciembre_2025_shp\Munpio.shp"
    gdf = gpd.read_file(RUTA_SHP)
    
    # ── 2. Reproyectar a WGS84 ────────────────────────────────────────────────
    gdf = gdf.to_crs("EPSG:4326")
    
    # ── 3. Filtrar por departamento ───────────────────────────────────────────
    depto_upper = nombre_departamento.strip().upper()
    gdf_depto = gdf[gdf["Depto"].str.upper().str.strip() == depto_upper]
    
    if gdf_depto.empty:
        deptos_disponibles = sorted(gdf["Depto"].str.title().unique())
        raise ValueError(
            f"❌ Departamento '{nombre_departamento}' no encontrado.\n"
            f"✅ Departamentos disponibles:\n{deptos_disponibles}"
        )
    
    # ── 4. GeoJSON y centro del mapa ──────────────────────────────────────────
    geojson = gdf_depto.__geo_interface__
    
    centro = {
        "lat": gdf_depto.geometry.centroid.y.mean(),
        "lon": gdf_depto.geometry.centroid.x.mean(),
    }
    
    # ── 5. Mapa coroplético ───────────────────────────────────────────────────
    fig = px.choropleth_mapbox(
        gdf_depto,
        geojson=geojson,
        locations=gdf_depto.index,
        color="MpArea",
        hover_name="MpNombre",
        hover_data={"MpArea": ":.2f"},
        color_continuous_scale="Viridis",
        mapbox_style="open-street-map",
        zoom=7,
        center=centro,
        opacity=0.65,
        labels={"MpArea": "Área (km²)"},
        title=f"Municipios de {nombre_departamento.title()} — Diciembre 2025",
    )
    
    fig.update_layout(
        margin={"r": 0, "t": 50, "l": 0, "b": 0},
        coloraxis_colorbar=dict(
            title="Área (km²)",
            thickness=15,
            len=0.6,
        ),
    )
    
    fig.show()


# ── USO ───────────────────────────────────────────────────────────────────────
mapa_departamento("Atlántico")
```
